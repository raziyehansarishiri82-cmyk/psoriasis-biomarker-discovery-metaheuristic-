```python
#!/usr/bin/env python3
"""
fetch_genes_mygene.py
ورودی: لیست سیمبل‌های ژنی (human). خروجی: JSON و CSV با نام کامل، توضیحات و شناسه‌ها.
نیازمند: pip install mygene
"""

import mygene
import csv
import json

# 🧬 لیست ژن‌ها
GENES = [
    "TNFRSF10C","MCOLN2","KYNU","CD47","TBX5-AS1","TEAD4","FCRLA","MDGA2","ZAN",
    "DNAAF4","C2orf74-DT","PDLIM7","PRRX1","FXR2","MRTFB","FZR1","MCCC2","RPLP2",
    "GSDMD","LINC00334","PEX10","CASTOR3P","VSTM2L","PPWD1","SOWAHC","PARD6B",
    "TNPO3","PLCD3","HSD17B14","SMARCD2","ADGRL3","H6PD","PRKAA2","GCLC","FN1"
]

# 🧠 ساخت آبجکت MyGene
mg = mygene.MyGeneInfo()

# مشخص کردن فیلدهای مورد نیاز
fields = "symbol,name,summary,entrezgene,ensembl,uniprot,alias,other_names,genomic_pos"

print("🔍 در حال دریافت اطلاعات از MyGene.info ...")
results = mg.querymany(GENES, scopes='symbol', fields=fields, species='human', as_dataframe=False)

# نرمال‌سازی خروجی
out_list = []
for r in results:
    entry = {
        "query": r.get('query'),
        "found": not r.get('notfound', False),
        "symbol": r.get('symbol'),
        "name": r.get('name'),
        "summary": r.get('summary'),
        "entrezgene": r.get('entrezgene'),
        "ensembl": r.get('ensembl'),
        "uniprot": r.get('uniprot'),
        "alias": r.get('alias') or r.get('other_names'),
        "genomic_pos": r.get('genomic_pos')
    }
    out_list.append(entry)

# ذخیره JSON
with open("genes_mygene.json", "w", encoding="utf-8") as f:
    json.dump(out_list, f, ensure_ascii=False, indent=2)

# ذخیره CSV
with open("genes_mygene.csv", "w", newline='', encoding="utf-8") as csvfile:
    fieldnames = ["query","found","symbol","name","summary","entrezgene","ensembl","uniprot","alias","genomic_pos"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for e in out_list:
        row = {
            k: (
                json.dumps(e[k], ensure_ascii=False)
                if isinstance(e[k], (dict, list))
                else ("" if e[k] is None else str(e[k]))
            )
            for k in fieldnames
        }
        writer.writerow(row)

# ✅ نمایش خلاصه‌ای از نتایج در ترمینال
print("\n📊 خلاصه‌ی نتایج:")
for item in out_list:
    if item["found"]:
        print(f"✅ {item['symbol']} - {item['name']}")
    else:
        print(f"❌ {item['query']} یافت نشد.")

print("\n💾 فایل‌های زیر ساخته شدند:")
print(" - genes_mygene.json")
print(" - genes_mygene.csv")
```
